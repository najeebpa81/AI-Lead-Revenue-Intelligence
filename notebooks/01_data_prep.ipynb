# 01 — Data audit and partitioning
Original project: B2B Commercial Lead Scoring Engine. This corrected version separates raw records **before** fitting preprocessing and excludes potentially post-outcome status columns.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from lead_scoring_utils import *

print("Dataset path:", SOURCE)
print("Dataset exists:", SOURCE.is_file())
```

```python

raw=pd.read_csv(SOURCE)
print('Shape:',raw.shape)
print('Duplicates:',raw.duplicated().sum())
print('Target distribution:',raw['Converted'].value_counts().to_dict())
display(raw.isna().mean().sort_values(ascending=False).head(15).to_frame('Missing fraction'))
```

```python
train,val,test=load_split()
print('Train/validation/test:',len(train),len(val),len(test))
print('Conversion rates:',*[round(x['Converted'].mean(),4) for x in (train,val,test)])
for name,part in [('train',train),('val',val),('test',test)]:
    part.to_csv(BASE / "data" / "processed" / f"{name}_raw.csv", index=False)
print('Saved raw partitions; no preprocessing fitted yet.')
```
